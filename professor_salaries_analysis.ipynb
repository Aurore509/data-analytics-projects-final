# Analysis of College Teacher Salaries

## Objective
The goal of this project is to analyze how college title and institution influence professor salaries using descriptive statistics and visualization.


```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("ct-professor-salaries.csv")
df.head()
```

```python
df.columns
```

```python
df.isnull().sum()
```

```python
df_clean = df.dropna(subset=["Average All Pay"])
df_clean.shape
```

```python
group1 = df_clean[df_clean["College"] == "Central Connecticut State University"]
```

```python
group2 = df_clean[df_clean["College"] == "University of Connecticut"]
```

```python
group1["Average All Pay"].describe()
```

```python
group2["Average All Pay"].describe()
```

```python
group1["Average All Pay"].mean(), group2["Average All Pay"].mean()
```

```python
import matplotlib.pyplot as plt

colleges = [ "Central Connecticut State University", "University of Connecticut" ]
means = [
    group1["Average All Pay"].mean(),
    group2["Average All Pay"].mean()
]

plt.bar(colleges, means)
plt.title("Average All Pay: CCSU vs UConn")
plt.xlabel("College")
plt.ylabel("Average All Pay")
plt.show()
```

## Conclusion

In this analysis, University of Connecticut faculty have a higher average overall pay than those at Central Connecticut State University.
This suggests that larger or research-focused institutions may offer higher compensation.
However, this dataset may be biased toward public reporting practices or specific job categories, and may not capture all forms of compensation.
Future analysis should consider discipline, rank differences, and cost of living to better understand salary variations.
